```python
# ================================
# VALIDATE 52-WEEK RANGE FORMAT
# ================================

range_validation = []

for index, value in df["52 Weeks Low - High"].items():

    # Convert the value to text
    text_value = str(value).strip()

    # Split using the hyphen
    parts = text_value.split("-")

    # A valid range should contain exactly two parts
    is_valid = len(parts) == 2

    range_validation.append({
        "Row": index,
        "Ticker": df.loc[index, "TY"],
        "Raw Value": value,
        "Number of Parts": len(parts),
        "Format Valid": is_valid
    })

range_validation_df = pd.DataFrame(range_validation)

range_validation_df[
    ~range_validation_df["Format Valid"]
]
```

```python
# ================================
# INSPECT EXCEL FORMULAS
# ================================

formula_records = []

for row in range(2, ws_formula.max_row + 1):

    ticker = ws_formula.cell(row=row, column=1).value

    for column in range(1, ws_formula.max_column + 1):

        cell = ws_formula.cell(row=row, column=column)

        # Check whether the Excel cell contains a formula
        if isinstance(cell.value, str) and cell.value.startswith("="):

            formula_records.append({
                "Excel Row": row,
                "Ticker": ticker,
                "Column": ws_formula.cell(
                    row=1,
                    column=column
                ).value,
                "Formula": cell.value,
                "Calculated Value": ws_values.cell(
                    row=row,
                    column=column
                ).value
            })

formula_df = pd.DataFrame(formula_records)

formula_df.head(20)
```

```python
# ================================
# FORMULA COUNT
# ================================

print("Total formula cells:", len(formula_df))

print(
    "Columns containing formulas:",
    formula_df["Column"].nunique()
)

formula_df["Column"].value_counts()
```

```python
# ================================
# DETECT EXCEL FORMULA ERRORS
# ================================

excel_error_patterns = [
    "#REF!",
    "#DIV/0!",
    "#VALUE!",
    "#NAME?",
    "#N/A",
    "#NUM!",
    "#NULL!"
]

formula_errors = formula_df[
    formula_df["Formula"].astype(str).apply(
        lambda x: any(error in x for error in excel_error_patterns)
    )
]

formula_errors
```

```python
# ================================
# CHECK CALCULATED VALUES
# ================================

calculated_value_errors = formula_df[
    formula_df["Calculated Value"].astype(str).isin(
        excel_error_patterns
    )
]

calculated_value_errors
```

```python
# ================================
# FORMULA DISTRIBUTION BY COLUMN
# ================================

formula_by_column = (
    formula_df
    .groupby("Column")
    .size()
    .sort_values(ascending=False)
)

formula_by_column
```

```python
# ================================
# INITIAL RAW / DERIVED CLASSIFICATION
# ================================

derived_columns = [
    "MARKET CAPITAL(ml)",
    "EPS IN LAST 12 MONTHS",
    "P/E BASED ON LAST ANNUAL REPORT",
    "P/E BASED ON LAST 12 MONTHS"
]

raw_columns = [
    column
    for column in df.columns
    if column not in derived_columns
]

feature_classification = pd.DataFrame({
    "Column": df.columns,
    "Classification": [
        "Derived / calculated"
        if column in derived_columns
        else "Raw / supplied"
        for column in df.columns
    ]
})

feature_classification
```

```python
# ================================
# VALIDATE MARKET CAPITALIZATION
# ================================

df["Calculated Market Cap"] = (
    df["TOTAL SHARES"] *
    df["PRICE"] /
    1_000_000
)

df["Market Cap Difference"] = (
    df["MARKET CAPITAL(ml)"] -
    df["Calculated Market Cap"]
)

market_cap_validation = df[
    [
        "TY",
        "MARKET CAPITAL(ml)",
        "Calculated Market Cap",
        "Market Cap Difference"
    ]
]

market_cap_validation.head(10)
```

```python
# ================================
# IDENTIFY MARKET CAP DIFFERENCES
# ================================

market_cap_problems = market_cap_validation[
    market_cap_validation["Market Cap Difference"].abs() > 0.01
]

market_cap_problems
```

```python
# ================================
# VALIDATE LTM EPS
# ================================

df["Calculated LTM EPS"] = (
    df["2H2024(JUL-DEC)"] +
    df["1H2025(JAN-JUNE)"]
)

df["EPS Difference"] = (
    df["EPS IN LAST 12 MONTHS"] -
    df["Calculated LTM EPS"]
)

eps_validation = df[
    [
        "TY",
        "EPS IN LAST 12 MONTHS",
        "Calculated LTM EPS",
        "EPS Difference"
    ]
]

eps_validation
```

```python
# ================================
# IDENTIFY EPS DIFFERENCES
# ================================

eps_problems = eps_validation[
    eps_validation["EPS Difference"].abs() > 0.0001
]

eps_problems
```

```python
# ================================
# VALIDATE LTM P/E
# ================================

df["Calculated LTM P/E"] = (
    df["PRICE"] /
    df["EPS IN LAST 12 MONTHS"]
) * 100

df["P/E Difference"] = (
    df["P/E BASED ON LAST 12 MONTHS"] -
    df["Calculated LTM P/E"]
)

pe_validation = df[
    [
        "TY",
        "P/E BASED ON LAST 12 MONTHS",
        "Calculated LTM P/E",
        "P/E Difference"
    ]
]

pe_validation.head(10)
```

```python
# ================================
# VALIDATE LTM P/E
# ================================

df["Calculated LTM P/E"] = (
    df["PRICE"] /
    df["EPS IN LAST 12 MONTHS"]
) * 100

df["P/E Difference"] = (
    df["P/E BASED ON LAST 12 MONTHS"] -
    df["Calculated LTM P/E"]
)

pe_validation = df[
    [
        "TY",
        "P/E BASED ON LAST 12 MONTHS",
        "Calculated LTM P/E",
        "P/E Difference"
    ]
]

pe_validation.head(10)
```

```python
# ============================================
# IDENTIFY ALL LTM P/E CALCULATION PROBLEMS
# ============================================

# Find companies where the Excel P/E differs
# from our independently calculated P/E.
#
# We use a small tolerance (0.01) because
# tiny floating-point differences are acceptable.

pe_problems = pe_validation[
    pe_validation["P/E Difference"].abs() > 0.01
]

# Display all problematic records
pe_problems
```
